```python
def init_para():
    W1=np.random.rand(10,784);
    b1=np.random.rand(10,1);
    w2=np.random.rand(10,10);
    b2=np.random.rand(10,1);
    return W1,b1,w2,b2
def relu(z):
    return np.maximum(0,z)
def softmax(z):
    exp_z=np.exp(z)
    return exp_z/np.sum(exp_z,axis=0,keepdims=True)
def forward_prop(X,W1,b1,w2,b2):
    Z1=np.dot(W1,X)+b1
    A1=relu(Z1)
    Z2=np.dot(w2,A1)+b2
    A2=softmax(Z2)
    return Z1,A1,Z2,A2
```
